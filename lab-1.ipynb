```python
#TASK 1
# Greedy Multi-Armed Bandit using Arithmetic Mean

# Number of machines
n = int(input("Enter number of machines: "))

machines = {}
q_values = {}

# Input rewards for each machine
for i in range(n):
    machine = input(f"Enter machine name {i + 1}: ")

    rewards = list(map(float, input(
        f"Enter rewards for Machine {machine} separated by spaces: "
    ).split()))

    machines[machine] = rewards

# Calculate Q(a)
for machine, rewards in machines.items():
    q_value = sum(rewards) / len(rewards)
    q_values[machine] = q_value

# Display results
print("\n--- Estimated Q(a) Values ---")

for machine, rewards in machines.items():
    print(f"\nMachine {machine}")

    # Display integers without .0
    display_rewards = [
        int(r) if r.is_integer() else r for r in rewards
    ]

    print(f"Rewards: {display_rewards}")
    print(f"Q({machine}) = {q_values[machine]:.2f}")

# Greedy selection
selected_machine = max(q_values, key=q_values.get)

print(f"\nGreedy Selected Machine: {selected_machine}")
```

```python
# Greedy Multi-Armed Bandit
# Incremental Update Method

# Number of machines
n = int(input("Enter number of machines: "))

Q = {}
N = {}
rewards = {}

# Task 1: Initialize
for i in range(n):
    machine = input("Enter machine name: ")
    reward_list = list(map(int, input("Enter rewards: ").split()))

    rewards[machine] = reward_list
    Q[machine] = 0
    N[machine] = 0

print("\n--- Incremental Updates ---")

# Task 2: Update after every play
for machine in rewards:

    print(f"\nMachine {machine}")

    for R in rewards[machine]:

        # Increment number of plays
        N[machine] += 1

        # Incremental update formula
        Q[machine] = Q[machine] + (R - Q[machine]) / N[machine]

        print(
            f"Reward = {R}, "
            f"N({machine}) = {N[machine]}, "
            f"Q({machine}) = {Q[machine]:.2f}"
        )

# Task 3: Display final Q values
print("\n--- Final Estimated Values ---")

for machine in Q:
    print(f"Machine {machine}")
    print(f"Rewards: {rewards[machine]}")
    print(f"Q({machine}) = {Q[machine]:.2f}")

# Greedy selection
selected_machine = max(Q, key=Q.get)

print("\nGreedy Selected Machine:", selected_machine)
```
