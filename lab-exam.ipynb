```python
#Task-1
states = ['S0', 'S1']
terminal_states = ['W', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.85
theta = 0.000001

transitions = {
    'S0': {
        'Up': ('S0', -2),
        'Down': ('S1', -2),
        'Left': ('S0', -2),
        'Right': ('S0', -2)
    },
    'S1': {
        'Up': ('S0', -2),
        'Down': ('S1', -2),
        'Left': ('S1', -2),
        'Right': ('G', 15)
    }
}

print("Non-terminal States:", states)
print("Terminal Outcomes:", terminal_states)
print("Action Space:", actions)
print("Discount Factor:", gamma)
print("Normal Movement Reward: -2")
print("Wall Reward: -2")
print("Goal Reward: +15")
print("Transition Mapping: 8 state-action pairs defined")
```

```python
#Task-2
print("Transition Mapping:")

for state in states:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, "|", action, "|", next_state, "|", reward)

print("All state-action pairs are defined:",
      all(action in transitions[state]
          for state in states for action in actions))
```

```python
#Task-3
V = {'S0': 0.0, 'S1': 0.0, 'G': 0.0}

print("Initial State Values:")
for state in V:
    print("V(" + state + ") =", format(V[state], ".4f"))
```

```python
#Task-4
V = {'S0': 0.0, 'S1': 0.0, 'G': 0.0}

iteration = 0
print("Iteration 0: V(S0)=0.0000, V(S1)=0.0000")

while True:
    Vnew = V.copy()
    delta = 0

    for state in states:
        values = []

        for action in actions:
            next_state, reward = transitions[state][action]
            values.append(reward + gamma * V[next_state])

        Vnew[state] = max(values)

    delta = max(abs(Vnew[s] - V[s]) for s in states)
    iteration += 1

    print("Iteration", iteration,
          ": V(S0)=", format(Vnew['S0'], ".4f"),
          ", V(S1)=", format(Vnew['S1'], ".4f"),
          ", Delta=", format(delta, ".6f"))

    V = Vnew.copy()

    if delta < theta:
        break

print("Converged: True")
print("Number of iterations:", iteration)
```

```python
#Task-5
Q = {}

for state in states:
    Q[state] = {}

    for action in actions:
        next_state, reward = transitions[state][action]
        Q[state][action] = reward + gamma * V[next_state]

        print("Q*(" + state + ", " + action + ") =",
              format(Q[state][action], ".4f"))
```

```python
#Task-6
print("Optimal Policy:")

policy = {}

for state in states:
    best_value = max(Q[state].values())

    best_actions = [
        action for action in actions
        if Q[state][action] == best_value
    ]

    policy[state] = best_actions
    print(state + ":", ", ".join(best_actions))

print("\nBellman Optimality Verification:")

verification = True

for state in states:
    max_q = max(Q[state].values())

    print("V*(" + state + ") =",
          format(V[state], ".4f"),
          "; max Q*(" + state + ",a) =",
          format(max_q, ".4f"))

    if abs(V[state] - max_q) > theta:
        verification = False

print("Verification:", verification)
```
